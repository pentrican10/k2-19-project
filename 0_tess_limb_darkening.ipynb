```python
import numpy as np
from ldtk import LDPSetCreator, BoxcarFilter
```

```python
# Define TESS bandpass wavelengths (example, adjust as needed)
tess_bandpass = BoxcarFilter('TESS', 600, 1000)

# Define filters with TESS bandpass
filters = [tess_bandpass]
```

```python
### params for K2-19 (Petigura et al 2020)
teff = 5322 # [K]
teff_err = 100
logg = 4.51 # [dex]
logg_err = 0.08
metallicity = 0.06 # [dex]
metallicity_err = 0.05
```

```python
# Set up LDPSetCreator with star parameters and filters
sc = LDPSetCreator(teff=(teff, teff_err),
                   logg=(logg, logg_err),
                   z=(metallicity, metallicity_err),
                   filters=filters)
```

```python
# Create limb darkening profiles
ps = sc.create_profiles()
```

```python
# Estimate quadratic law coefficients
cq, eq = ps.coeffs_qd(do_mc=True)
```

```python
# Print them out
for i, (coeffs, errors) in enumerate(zip(cq, eq)):
    print(f"Filter {i}:")
    print(f"  u1 = {coeffs[0]:.4f} ± {errors[0]:.4f}")
    print(f"  u2 = {coeffs[1]:.4f} ± {errors[1]:.4f}")
```

```python
# Calculate likelihood for a set of coefficients (optional)
lnlike = ps.lnlike_qd([[0.45, 0.15], [0.35, 0.10], [0.25, 0.05]])
```
